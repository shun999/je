```python
df = df.iloc[:240]
```

```python
df_high_all = df[df["subject"].isin([1, 2, 5, 6, 9])]
df_low_all = df[df["subject"].isin([3, 4, 7, 10, 11])]
```

```python
df_high_all = df_high_all[df_high_all["q_category"]==3].reset_index(drop=True)
df_low_all = df_low_all[df_low_all["q_category"]==3].reset_index(drop=True)
```

```python
df_high_all.columns
```

```python
for i, column in enumerate(df_high_all.columns):
    target = column
    df_compare = pd.DataFrame({"High": df_high_all[target],
                            "Low": df_low_all[target]})

    import matplotlib.pyplot as plt
    import seaborn as sns

    plt.rcParams['font.family'] = 'MS Gothic'  # または 'Meiryo'
    plt.figure(figsize=(4,3))

    sns.boxplot(data=df_compare, palette="Set3", width=0.3,
                boxprops=dict(linewidth=2, ),        # 箱boxprops=dict(linewidth=2, edgecolor='black'),      # 箱の線
                medianprops=dict(linewidth=2, color='red'),         # 中央値線
                whiskerprops=dict(linewidth=2, color='black'),      # ひげ
                capprops=dict(linewidth=2, color='black')           # ひげの端
)
    
    # plt.title(f'High vs Low {target}', fontweight='bold')
    plt.ylabel(f"Value for\n{target}", fontweight='bold', fontsize=17)
    # plt.xlabel(target, fontweight='bold', fontsize=17)
    plt.tick_params(labelsize=12, width=2)  # 文字サイズ・目盛り線の太さ
    for label in plt.gca().get_xticklabels() + plt.gca().get_yticklabels():
        label.set_fontweight('bold')         # 文字を太字に
    plt.show()

    df_compare=df_compare.dropna()

    from scipy.stats import brunnermunzel

    t_stat, p_value = brunnermunzel(df_compare["High"], df_compare["Low"])

    print("t値 :", t_stat)
    print("p値 :", p_value)
```

```python
from scipy.stats import permutation_test
import numpy as np

for i, column in enumerate(df_high_all.columns):
    target = column
    df_compare = pd.DataFrame({"High": df_high_all[target],
                            "Low": df_low_all[target]})

    import matplotlib.pyplot as plt
    import seaborn as sns

    plt.rcParams['font.family'] = 'MS Gothic'  # または 'Meiryo'

    sns.boxplot(data=df_compare, palette="Set3")
    plt.title(f'High vs Low ')
    plt.ylabel("score")
    plt.xlabel(target)
    plt.show()

    df_compare=df_compare.dropna()

    def statistic(a, b):
        return a.mean() - b.mean()

    result = permutation_test(
        (df_compare["High"], df_compare["Low"]),
        statistic,
        n_resamples=5000,
        alternative='two-sided'
    )

    print(result.statistic)
    print(result.pvalue)
```
