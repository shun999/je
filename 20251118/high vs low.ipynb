```python
for i, column in enumerate(df_high_all.columns):
    target = column
    df_compare = pd.DataFrame({"High": df_high_all[target],
                            "Low": df_low_all[target]})

    import matplotlib.pyplot as plt
    import seaborn as sns

    plt.rcParams['font.family'] = 'MS Gothic'  # または 'Meiryo'

    sns.boxplot(data=df_compare, palette="Set3")
    plt.title(f'High vs Low ')
    plt.ylabel("score")
    plt.xlabel(target)
    plt.show()

    # df_compare["High"]=df_compare["High"].dropna()
    # df_compare["Low"]=df_compare["Low"].dropna()

    a = df_compare["High"].dropna().to_numpy()
    b = df_compare["Low"].dropna().to_numpy()

    # print(df_compare["High"].describe())
    # print(df_compare["Low"].describe())

    # print(df_compare["High"].isna().sum(), df_compare["Low"].isna().sum())

    from scipy.stats import brunnermunzel

    t_stat, p_value = brunnermunzel(a, b)
    # t_stat, p_value = brunnermunzel(df_compare["High"], df_compare["Low"])

    print("統計量 :", t_stat)
    print("p値 :", p_value)
```
