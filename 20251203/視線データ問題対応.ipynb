```python
import pandas as pd
```

```python
df = pd.read_csv(r"F:\Je respire\20251203\視線データ\tobii_data_20251126_153049_iwasaki.csv")
```

```python
idx = df.index[df["Marker"].notna()]
print(idx)
```

```python
from sklearn.preprocessing import StandardScaler

# 数値列だけ抽出
numeric_df = df.select_dtypes(include=["number"])

# 標準化
scaler = StandardScaler()
scaled_values = scaler.fit_transform(numeric_df)

# 結果を元のデータフレームに反映
df_result = df.copy()
df_result[numeric_df.columns] = scaled_values

print(df_result)
```

```python
df1 = df_result.iloc[idx[0]:idx[1]]
df2 = df_result.iloc[idx[2]:idx[3]]
df3 = df_result.iloc[idx[4]:idx[5]]
df4 = df_result.iloc[idx[6]:idx[7]]
df5 = df_result.iloc[idx[8]:idx[9]]
df6 = df_result.iloc[idx[10]:idx[11]]
```

```python
with pd.ExcelWriter("sample.xlsx") as writer:
    df1.to_excel(writer, sheet_name="N1", index=False)
    df2.to_excel(writer, sheet_name="N2", index=False)
    df3.to_excel(writer, sheet_name="N3", index=False)
    df4.to_excel(writer, sheet_name="N4", index=False)
    df5.to_excel(writer, sheet_name="N5", index=False)
    df6.to_excel(writer, sheet_name="N6", index=False)
```
