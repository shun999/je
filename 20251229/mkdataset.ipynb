```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\human1\OneDrive - 埼玉大学\デスクトップ\12.29\視線データ\1. kubota.csv")
```

```python
# Marker列の値が存在するインデックスを取得
marker_indices = df[df['Marker'].notna()].index
print(f"Marker列に値が存在するインデックス数: {len(marker_indices)}")
print(f"インデックス: {marker_indices.tolist()}")
```

```python
df.iloc[marker_indices]
```

```python
# インデックス間のデータをExcelの各シートに出力
marker_indices_list = marker_indices.tolist()

# Excelファイルのパスを指定
excel_path = r"C:\Users\human1\OneDrive - 埼玉大学\デスクトップ\12.29\視線データ\1. kubota_segments.xlsx"

with pd.ExcelWriter(excel_path, engine='openpyxl') as writer:
    # 最初のセグメント（最初のマーカーまで）
    if len(marker_indices_list) > 0:
        start_idx = 0
        end_idx = marker_indices_list[0]
        segment_df = df.iloc[start_idx:end_idx+1]  # end_idx+1でマーカー行も含める
        sheet_name = f"Segment_0_to_{end_idx}"
        segment_df.to_excel(writer, sheet_name=sheet_name, index=False)
        print(f"シート '{sheet_name}': {len(segment_df)}行を出力")
    
    # 中間のセグメント（マーカー間のデータ）
    for i in range(len(marker_indices_list) - 1):
        start_idx = marker_indices_list[i]
        end_idx = marker_indices_list[i + 1]
        segment_df = df.iloc[start_idx:end_idx+1]  # end_idx+1で次のマーカー行も含める
        sheet_name = f"Segment_{start_idx}_to_{end_idx}"
        segment_df.to_excel(writer, sheet_name=sheet_name, index=False)
        print(f"シート '{sheet_name}': {len(segment_df)}行を出力")
    
    # 最後のセグメント（最後のマーカーから最後まで）
    if len(marker_indices_list) > 0:
        start_idx = marker_indices_list[-1]
        end_idx = len(df) - 1
        segment_df = df.iloc[start_idx:end_idx+1]
        sheet_name = f"Segment_{start_idx}_to_end"
        segment_df.to_excel(writer, sheet_name=sheet_name, index=False)
        print(f"シート '{sheet_name}': {len(segment_df)}行を出力")

print(f"\nExcelファイルに出力完了: {excel_path}")
```

```python
# Marker列の値を確認
print("Marker列の値:")
print(df.loc[marker_indices, 'Marker'].tolist())
```

```python
# Marker列で"s"から"e"までのデータを抽出
# "s"で始まるマーカーのインデックスを取得
s_indices = df[df['Marker'].notna() & (df['Marker'].astype(str).str.startswith('s'))].index
# "e"で終わるマーカーのインデックスを取得
e_indices = df[df['Marker'].notna() & (df['Marker'].astype(str).str.endswith('e'))].index

print(f"'s'で始まるマーカーのインデックス: {s_indices.tolist()}")
print(f"'e'で終わるマーカーのインデックス: {e_indices.tolist()}")

# 最初の"s"から最後の"e"までのデータを抽出
if len(s_indices) > 0 and len(e_indices) > 0:
    start_idx = s_indices[0]
    end_idx = e_indices[-1]
    
    # データを抽出（end_idx+1で最後の"e"の行も含める）
    segment_df = df.iloc[start_idx:end_idx+1]
    
    # Excelファイルに出力
    excel_path = r"C:\Users\human1\OneDrive - 埼玉大学\デスクトップ\12.29\視線データ\1. kubota_s_to_e.xlsx"
    segment_df.to_excel(excel_path, sheet_name='s_to_e', index=False, engine='openpyxl')
    
    print(f"\n抽出したデータ: {len(segment_df)}行")
    print(f"開始インデックス: {start_idx}, 終了インデックス: {end_idx}")
    print(f"Excelファイルに出力完了: {excel_path}")
else:
    print("'s'で始まるマーカーまたは'e'で終わるマーカーが見つかりませんでした。")
```

```python
# "s1"から"e1"、"s2"から"e2"のようにペアごとにデータを抽出してExcelの各シートに出力
marker_df = df.loc[marker_indices, 'Marker'].reset_index()
marker_df.columns = ['index', 'marker']

# Excelファイルのパスを指定
excel_path = r"C:\Users\human1\OneDrive - 埼玉大学\デスクトップ\12.29\視線データ\1. kubota_pairs.xlsx"

with pd.ExcelWriter(excel_path, engine='openpyxl') as writer:
    # s1-e1, s2-e2などのペアを抽出
    for i in range(1, 7):  # s1-e1からs6-e6まで
        s_marker = f's{i}'
        e_marker = f'e{i}'
        
        # sとeのインデックスを取得
        s_idx = marker_df[marker_df['marker'] == s_marker]['index'].values
        e_idx = marker_df[marker_df['marker'] == e_marker]['index'].values
        
        if len(s_idx) > 0 and len(e_idx) > 0:
            start_idx = s_idx[0]
            end_idx = e_idx[0]
            
            # データを抽出（end_idx+1でeの行も含める）
            segment_df = df.iloc[start_idx:end_idx+1]
            
            # シート名（Excelのシート名は31文字まで、一部の特殊文字は使用不可）
            sheet_name = f"{s_marker}_to_{e_marker}"
            segment_df.to_excel(writer, sheet_name=sheet_name, index=False)
            print(f"シート '{sheet_name}': {len(segment_df)}行を出力 (インデックス {start_idx} から {end_idx})")
        else:
            print(f"警告: {s_marker} または {e_marker} が見つかりませんでした")

print(f"\nExcelファイルに出力完了: {excel_path}")
```
